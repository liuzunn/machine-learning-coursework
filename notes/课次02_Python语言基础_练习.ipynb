## 第 0 步：环境自检
确认 sklearn 能导入。报错就回头重装环境（见《课前教程_Python环境安装指南》）。

```python
# TODO: 导入 sklearn 和 sklearn.datasets 里的 load_iris，并打印 sklearn 版本号
import sklearn
from sklearn.datasets import load_iris

print("sklearn 版本：", sklearn.__version__)
```

## 第 1 步：变量与内置类型
Python 里**一切变量皆是对象**，变量存的是对象的内存地址。用 `type()` 看看每个变量的类型。

```python
# TODO: 用 load_iris() 读数据，把 iris.data 转成列表赋给 X
iris = load_iris()
X = iris.data.tolist()

# TODO: 定义 n = 样本数、first = 第一朵花
n = len(X)
first = X[0]

# TODO: 用 for 循环累加第 0 列，算出花萼长度平均值 avg（提示：先 total = 0，再循环累加）
total = 0.0
for row in X:
    total = total + row[0]
avg = total / n

# TODO: 用 print 和 type() 打印这三个变量的值和类型
print("n     =", n, "    类型:", type(n))
print("first =", first, "类型:", type(first))
print("avg   =", avg, "类型:", type(avg))
```

## 第 2 步：索引与切片
切片语法 `[start: end: step]`：end **不包含**，step 为负表示**反向**。切片得到的是**新列表**（浅拷贝）。

```python
# TODO: 取出 X[0] 赋给 s，打印：第 1 个元素 s[0]、最后一个 s[-1]
s = X[0]
print("第 1 个元素 s[0] =", s[0])
print("最后一个   s[-1] =", s[-1])

# TODO: 打印中间两个元素 s[1:3]、逆序 s[::-1]、隔一个取一个 s[::2]
print("中间两个 s[1:3]  =", s[1:3])
print("逆序     s[::-1] =", s[::-1])
print("隔一个   s[::2]  =", s[::2])

# TODO: 打印前 3 朵花 X[:3]
print("前 3 朵花 X[:3]  =", X[:3])
```

## 第 3 步：循环 + 判断
`for` 遍历列表，`if` 做判断。统计：有多少朵花的花瓣长度（第 3 列，下标 2）超过平均值？

```python
# TODO: 用 for 循环 + append 取出所有花的花瓣长度（每行下标 2）放进 petals
petals = []
for row in X:
    petals.append(row[2])

# TODO: 计算平均值 pavg（sum / len）
pavg = sum(petals) / len(petals)
print("花瓣长度平均值 pavg =", pavg)

# TODO: 用 for + if 统计 petals 中大于 pavg 的个数，打印结果
count = 0
for p in petals:
    if p > pavg:
        count = count + 1
print("花瓣长度大于平均值的有", count, "朵")
```

## 第 4 步：函数封装
把第 3 步的逻辑装进函数：`def` 定义、`return` 返回。注意**缩进就是函数体**。

```python
# TODO: 定义函数 count_above(data, col, threshold)，统计 data 中第 col 列大于 threshold 的行数并 return
def count_above(data, col, threshold):
    count = 0
    for row in data:
        if row[col] > threshold:
            count = count + 1
    return count

# TODO: 调用它统计“花瓣超平均”和“花萼长度超 6.0”的花朵数并打印
print("花瓣长度超平均的朵数  :", count_above(X, 2, pavg))
print("花萼长度超 6.0 的朵数 :", count_above(X, 0, 6.0))
```

## 挑战（可选）
用**切片 + 循环**（不许用 sort）回答：前 10 朵花里，花瓣最长的是第几朵？多长？

```python
# TODO: 用 X[:10] 切片 + 循环，找出花瓣长度（下标 2）最大的那朵
#       提示：自己用 i = 0; i = i + 1 计数，不许用 sort 和 enumerate
best_i = 0      # 目前最大值的下标（从 0 开始）
best_v = -1.0   # 目前见过的最大花瓣长度
i = 0
for row in X[:10]:
    if row[2] > best_v:
        best_v = row[2]
        best_i = i
    i = i + 1

# TODO: 打印序号（从 1 开始）和长度
print("前 10 朵花中花瓣最长的是第", best_i + 1, "朵，长度 =", best_v)
```
