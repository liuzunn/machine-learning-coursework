## 一、NumPy（对应 PPT 第 9–19 页）


### 1.1 创建数组（PPT 第 9–12 页）


```python
import numpy as np

# ① 由列表创建
a = np.array([1, 2, 3, 4, 5])
print(a, '| shape:', a.shape)

# ② arange：指定步长，不含终点
print(np.arange(0, 10, 2))        # [0 2 4 6 8]

# ③ linspace：指定个数，含终点
print(np.linspace(0, 10, 5))      # [ 0.   2.5  5.   7.5 10. ]
```

```python
# ④ 特殊矩阵
print(np.zeros((2, 3)))          # 全零
print(np.ones((2, 3)))           # 全一
print(np.identity(3))            # 单位阵
```

### 1.2 数组运算（PPT 第 13–14 页）


```python
x = np.array([1, 2, 3])

print(x * 2)          # 每个元素 ×2 → [2 4 6]
print(x + 10)         # 每个元素 +10 → [11 12 13]
print(x * x)          # 对应元素相乘 → [1 4 9]
print(x.sum())        # 6
```

### 1.3 元素访问与切片（PPT 第 15–17 页）


```python
b = np.array([[1, 2, 3],
              [4, 5, 6]])
print(b[0, 2])     # 第 0 行第 2 列 → 3
print(b[1, :])     # 第 1 行全部 → [4 5 6]
print(b[:, 1])     # 第 1 列全部 → [2 5]
```

```python
a = np.array([10, 20, 30, 40, 50])
print(a[1:4])      # 下标 1~3 → [20 30 40]（end 取不到）
print(a[::-1])     # 逆序 → [50 40 30 20 10]
print(a[::2])      # 隔一个取 → [10 30 50]
```

### 1.4 常用计算（PPT 第 18 页）


```python
d = np.array([[1, 2, 3], [4, 5, 6]])
print('sum  =', d.sum())        # 21
print('mean =', d.mean())       # 3.5
print('max  =', d.max())        # 6
print('min  =', d.min())        # 1
print('std  =', round(d.std(), 3))
print('按列 sum:', d.sum(axis=0))   # [5 7 9]
print('按行 sum:', d.sum(axis=1))   # [ 6 15]
```

## 二、Pandas（对应 PPT 第 20–31 页）


### 2.1 Series：一维数据（PPT 第 22 页）


```python
import pandas as pd

s = pd.Series([1, 3, 5, np.nan])
print(s)

s2 = pd.Series([85, 92, 78], index=['张三', '李四', '王五'])
print(s2['李四'])      # 可按标签取值 → 92
```

### 2.2 DataFrame：二维数据（PPT 第 23 页）


```python
# ① 字典创建（最常用）
df = pd.DataFrame({'姓名': ['张三', '李四', '王五'],
                   '年龄': [20, 26, 22],
                   '班级': ['A班', 'B班', 'A班']})
print(df)

print('shape:', df.shape)      # (3, 3) → 3 行 3 列
```

```python
# ② 列表创建（需指定列名）
df2 = pd.DataFrame([[1, 'a'], [2, 'b']], columns=['数字', '字母'])
print(df2)

# ③ 读文件（后续课次最常用）
# df3 = pd.read_csv('data.csv')
```

### 2.3 一维与二维的区别（PPT 第 24 页）


```python
print('Series 形状:', s2.shape)     # (3,)    → 一条
print('DataFrame 形状:', df.shape)  # (3, 3)  → 一张表
```

### 2.4 查看数据（PPT 第 25 页）


```python
print(df.head(2))         # 前 2 行
print(df.info())          # 结构与类型
print(df.describe())      # 数值列统计量
```

### 2.5 选择数据（PPT 第 26 页）


```python
print(df['姓名'])         # 选一列（Series）
print(df[['姓名', '年龄']])  # 选多列
print(df.loc[0])          # 按行标签取（第 0 行）
print(df.iloc[1])         # 按行位置取（第 1 行）
```

### 2.6 排序与统计（PPT 第 27–28 页）


```python
print(df.sort_values('年龄', ascending=False))   # 按年龄降序

print(df['年龄'].mean())    # 平均值
print(df['年龄'].max())     # 最大值
print(df.groupby('班级')['年龄'].mean())   # 按班级分组求平均
```

### 2.7 例题：商品销售分析（PPT 第 31 页）


```python
df_sale = pd.DataFrame({'商品': ['键盘', '鼠标', '显示器', '耳机', '音箱'],
                        '单价': [199, 89, 899, 299, 399],
                        '销量': [12, 35, 4, 9, 6]})
df_sale['销售额'] = df_sale['单价'] * df_sale['销量']
print(df_sale.sort_values('销售额', ascending=False))
```

## 三、Matplotlib（对应 PPT 第 32–41 页）


```python
import matplotlib.pyplot as plt
# plt.rcParams['font.sans-serif'] = ['SimHei']    # 中文正常显示
plt.rcParams['font.sans-serif'] = ['WenQuanYi Micro Hei']    # 中文正常显示
plt.rcParams['axes.unicode_minus'] = False

names = ['张三', '李四', '王五', '赵六', '钱七']
scores = [85, 92, 78, 90, 66]
```

### 3.1 基础：点图与折线图（PPT 第 33 页）


```python
fig, ax = plt.subplots(1, 2, figsize=(8, 3))
ax[0].scatter(range(5), scores)          # 点图：看关系
ax[0].set_title('成绩点图')
ax[1].plot(names, scores, marker='o')    # 折线图：看趋势
ax[1].set_title('成绩折线图')
plt.tight_layout()
plt.show()
```

### 3.2 例题：折线图 + 柱状图（PPT 第 41 页）


```python
fig, ax = plt.subplots(1, 2, figsize=(8, 3))
ax[0].plot(names, scores, marker='o', color='#0070C0')
ax[0].set_title('成绩折线图'); ax[0].set_ylabel('分数')
ax[1].bar(names, scores, color='#4472C4')
ax[1].set_title('成绩柱状图'); ax[1].set_ylabel('分数')
plt.tight_layout()
plt.show()
```

## 四、scikit-learn（对应 PPT 第 42–53 页）


```python
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split

X, y = load_iris(return_X_y=True)
print('样本数:', X.shape[0])   # 150
print('特征数:', X.shape[1])   # 4
print('类别数:', len(set(y)))  # 3

Xtr, Xte, ytr, yte = train_test_split(X, y, random_state=0)
knn = KNeighborsClassifier().fit(Xtr, ytr)
print('准确率:', knn.score(Xte, yte))   # 约 0.97
```
